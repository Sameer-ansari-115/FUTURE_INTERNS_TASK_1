```python
# monthly revenue trend
```

```python
data["Order Date"]=pd.to_datetime(data["Order Date"])
monthly_revenue=data.groupby(data["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_revenue.plot(figsize=(12,5),marker='o',linestyle="--")
plt.title("Monthly_Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
# top 10 selliing product
```

```python
top_product=data.groupby("Product Name")["Quantity"].sum().sort_values(ascending=False).head(10)
top_product.plot( kind="bar",figsize=(12,5))
plt.title("top 10 selliing product")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
# plt.xticks(rotation=75)
plt.show()
```

```python
# High-value categories or regions
```

```python
high_category =data.groupby("Category")["Sales"].sum().sort_values(ascending=False).head(10)
high_category.plot(kind="bar",figsize=(8,5))

plt.title("High-Value Categories by Revenue")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.show()
```

```python
# high value region
```

```python
high_region = (data.groupby("Region")["Sales"].sum().sort_values(ascending=False))
high_region.plot(kind="bar",figsize=(10,5))

plt.title("High-Value Regions by Revenue")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.xticks(rotation=0)
plt.show()
```

```python
category_profit = data.groupby("Category")["Profit"].sum().sort_values(ascending=False)
category_profit.plot(kind="bar",figsize=(10,5))

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()
```

```python
# Region by profit
```

```python
region_profit =data.groupby("Region")["Profit"].sum().sort_values(ascending=False)
region_profit.plot(kind="bar",figsize=(10,5))

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.show()

```

# Business Insights

1. Office Supplies is the highest revenue-generating category, generating total sales of approximately 719,047.

2. The West region is the highest revenue-generating region, contributing approximately 725,458 in sales.

3. Technology is the most profitable category, generating approximately 145,455 in profit.

4. November 2017 recorded the highest monthly revenue, with approximately 118,448 in sales.

5. Staples is the top-selling product by quantity, with 215 units sold.

6. The results show that revenue and profitability are not always driven by the same category, highlighting the importance of analyzing both sales and profit.

# Business Recommendations

1. Focus on the Office Supplies category by maintaining sufficient inventory and exploring opportunities to increase its sales further.

2. Continue investing in the West region through targeted marketing and customer retention strategies because it generates the highest revenue.

3. Increase focus on the Technology category because it generates the highest profit and can contribute significantly to business profitability.

4. Plan promotional campaigns and inventory strategies around high-demand periods such as November.

5. Ensure adequate stock availability for high-demand products such as Staples to avoid stockouts.

6. Evaluate categories using both revenue and profit metrics before making business decisions, as the highest revenue category may not always be the most profitable.

# Conclusion

This analysis provides an overview of business sales performance by examining revenue trends, product demand, category performance, regional performance, and profitability.

The analysis found that Office Supplies generated the highest revenue, while Technology was the most profitable category. The West region generated the highest revenue, and November 2017 recorded the highest monthly revenue. Staples was the top-selling product based on quantity sold.

These findings can help the business improve inventory planning, marketing strategies, regional focus, and profitability.
